# Seaborn Demo

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

## What is seaborn, and how does it relate to matplotlib?

Seaborn is a plotting library built **on top of** matplotlib - it is not a replacement for anything you already learned. Every seaborn function still returns a matplotlib `Axes` object (or, for a few multi-plot functions, a `FacetGrid`), so everything you know about `ax.set_xlabel()`, `ax.set_title()`, `fig.savefig()`, etc. still applies.

What seaborn actually adds:

    1. It works directly with DataFrames - you pass a `data=` argument and refer to columns by name, instead of pulling out raw lists/arrays yourself.
    2. It has better statistical defaults - confidence intervals, aggregation, and color palettes are handled for you.
    3. Several common statistical plots (distributions, categorical comparisons, pairwise relationships) that would take many lines of matplotlib code are a single function call.

Let's load the Ames housing data again and see this in action.

```python
ames = pd.read_csv("AmesHousing.csv")
ames = ames[["Lot Area", "Year Built", "Street", "Bldg Type", "SalePrice", "House Style", "Fireplaces", "Overall Qual", "Overall Cond"]]
ames.head()
```

## Same plot, less code

Here is a plot you already know how to build in matplotlib, and the seaborn equivalent, side by side.

```python
# matplotlib way - you already know this
fig, ax = plt.subplots()
ax.scatter(ames["Lot Area"], ames["SalePrice"])
ax.set_xlabel("Lot Area")
ax.set_ylabel("Sale Price")
```

```python
# seaborn way - pass the data frame once, then refer to columns by name
sns.scatterplot(data=ames, x="Lot Area", y="SalePrice")
```

Notice the pattern: `data=ames` tells seaborn which DataFrame to pull from, and `x=`/`y=` are just column name strings. This `data=` + column-name pattern shows up in almost every seaborn function, so it's worth getting comfortable with early.

## Distributions - `histplot` and `kdeplot`

In matplotlib, a histogram means picking bin counts/widths yourself. `sns.histplot()` handles this automatically, and can also overlay a smoothed density curve.

```python
sns.histplot(data=ames, x="SalePrice", kde=True)
```

## Categorical comparisons - `boxplot` and `violinplot`

Earlier, in the pandas demo, you compared `SalePrice` across `Overall Qual` groups using `groupby()` and looked at the means. Seaborn can show the whole distribution for each group, not just the mean, in a single line.

```python
sns.boxplot(data=ames, x="Overall Qual", y="SalePrice")
```

A violin plot shows the same grouped comparison, but draws the full shape of each group's distribution instead of just quartiles.

```python
sns.violinplot(data=ames, x="Overall Qual", y="SalePrice")
```

## Color-coding by category with `hue`

Adding a third, categorical variable to a plot in matplotlib means manually looping over groups and calling `.scatter()` once per group. In seaborn, it's one extra argument: `hue`.

```python
sns.scatterplot(data=ames, x="Lot Area", y="SalePrice", hue="Bldg Type")
```

## Pairwise relationships - `pairplot`

`sns.pairplot()` builds a full grid of scatter plots for every pair of numeric columns, with histograms along the diagonal - all in a single function call. This is a good example of something that would take a lot of manual subplot work in raw matplotlib.

```python
subset = ames[ames["Bldg Type"].isin(["1Fam", "TwnhsE"])] #subset categories to make plot less cluttered
sns.pairplot(subset[["Lot Area", "SalePrice", "Overall Qual", "Overall Cond", "Bldg Type"]], hue = "Bldg Type")
```

## Customizing seaborn plots with matplotlib

Since every seaborn function returns a matplotlib `Axes` object, you can capture it and keep using everything you already know for titles, labels, and text sizing.

```python
ax = sns.boxplot(data=ames, x="Overall Qual", y="SalePrice")
ax.set_title("Sale Price by Quality Rating", size=18)
ax.set_xlabel("Overall Quality", size=15)
ax.set_ylabel("Sale Price", size=15)
```

## A note on figure size

Most seaborn functions accept an `ax=` argument, which lets you draw onto an Axes you created yourself with `plt.subplots()` - this is the easiest way to control figure size or place a seaborn plot into a larger grid of subplots.

```python
fig, ax = plt.subplots(figsize=(8, 5))
sns.violinplot(data=ames, x="Overall Qual", y="SalePrice", ax=ax)
ax.set_title("Sale Price by Quality Rating", size=18)
```
