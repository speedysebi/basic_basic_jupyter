# CSC 2611 - Jupyter Introduction

This notebook is a short introduction to **Jupyter Notebook** and how it differs from working in PyCharm.

The goal is not to learn new Python today. The goal is to learn a new **workflow** for writing, running, documenting, and exploring Python code.

### Mental model
- **PyCharm:** build and organize software
- **Jupyter:** explore, analyze, visualize, and explain

You will often use both tools on the same project.

## 1. A notebook is made of cells

A notebook is not one program that must always run from top to bottom. It is made of individual **cells**.

There are two cell types we care about today:

1. **Code cells** - contain Python that can be executed.
2. **Markdown cells** - contain formatted notes, explanations, headings, equations, and other documentation.

Click the next cell and run it with **Shift + Enter**.

```python
x = 5
x * 2
```

Notice what happened:

- The cell executed.
- Jupyter displayed the value of the final expression directly underneath it.
- We did not need to write `print(x * 2)` just to see the result.

You **can** still use `print()` whenever you want.

```python
print(x * 2)
```

## 2. The notebook remembers state

Jupyter has a running Python process behind the notebook called the **kernel**.

When a cell runs, variables created in that cell remain available to later cells.

```python
x + 10
```

The cell above knows what `x` is because the earlier cell defining `x = 5` has already run.

That is powerful, but it is also one of the easiest ways to create confusing notebooks.

### Important rule
A healthy notebook should work when you:

**Restart the kernel -> Run all cells from top to bottom**

If a notebook only works because you ran cells in a strange order, it is not reproducible.

## 3. Execution order matters

The number shown to the left of a code cell tells you its execution order.

Try changing the value below and running this cell several times.

```python
y = 10
print("y =", y)
```

```python
print("y squared =", y ** 2)
```

If you restart the kernel and try to run only the cell that uses `y`, Python will not know what `y` is yet.

That is why **Restart + Run All** is such an important check before submitting or sharing a notebook.

## 4. Markdown lets us explain the analysis

This entire cell is Markdown rather than Python.

Markdown supports simple formatting:

- **Bold text**
- *Italic text*
- Lists
- `inline code`
- Headings

### Why this matters
When we start doing data analysis and machine learning, a notebook can contain:

1. The question we are trying to answer
2. The Python used to analyze the data
3. A graph or table of the result
4. Our interpretation of what the result means

That is one of Jupyter's biggest strengths.

## 5. A tiny analysis example

Here is a small set of measurements. We can calculate a result and immediately see it below the code.

```python
values = [3, 5, 4, 8, 10, 6]

average = sum(values) / len(values)
average
```

## 6. Visualizations appear inline

Jupyter is especially useful when code produces a graph or other visual output.

Do not worry about memorizing the Matplotlib syntax yet. The point of this example is to see the workflow.

```python
import matplotlib.pyplot as plt

plt.plot(values, marker="o")
plt.title("Example Measurements")
plt.xlabel("Measurement Number")
plt.ylabel("Value")
plt.show()
```

The graph stays directly next to the code and explanation that produced it.

This becomes extremely useful when we are exploring data or comparing machine-learning results.

## 7. Notebook workflow

A useful Jupyter workflow is:

**Question -> Code -> Result -> Interpretation**

For example:

### Question
What is the largest value in our list, and where does it occur?

```python
largest_value = max(values)
largest_index = values.index(largest_value)

print("Largest value:", largest_value)
print("Index:", largest_index)
```

### Interpretation
The largest measurement is **10**, and it occurs at index **4**.

This pattern of mixing explanation, executable code, and interpretation is what makes notebooks useful for technical analysis.
