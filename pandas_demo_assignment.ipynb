# Pandas Demo

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

## Reading in a data set

pandas has a built-in CSV reader (`pd.read_csv`), as well as readers for many other file types (Excel, JSON, SQL, etc.). Reading a file in gives you back a **DataFrame** - pandas' main data structure, basically a table with labeled rows and columns.

```python
#Note - parts of this tutorial are inspired by https://realpython.com/pandas-python-explore-dataset/

ames = pd.read_csv("AmesHousing.csv")
type(ames)
```

```python
print(len(ames)) #rows in data set
print(ames.shape) #tuple with (rows, cols)
```

## Getting a first look at the data

A few methods that are almost always worth running right after you load a new data set.

```python
ames.head()
```

```python
# we are missing some columns - we can set an option to see all columns and be able to scroll
pd.set_option("display.max.columns", None)
```

```python
ames.tail(10) # n defaults to 5 like for .head()
```

```python
# like numpy arrays, each column in pandas has a data type
# unlike numpy arrays, each column can have a different data type
ames.info()
```

```python
ames.describe()
```

## Selecting columns

Most real data sets have more columns than you actually need for a given task. To select a specific subset of columns, put a list of the column names in the selection brackets.

```python
# we probably do not need all 82 columns - lets select the columns we really care about
ames = ames[["Lot Area", "Year Built", "Street", "Bldg Type", "SalePrice", "House Style", "Fireplaces", "Overall Qual", "Overall Cond"]]
ames.head()
```

## Counting values

To determine how often specific values occur in a column, use the `.value_counts()` method on that column.

```python
print(ames["Overall Qual"].value_counts())
print()
print(ames["Street"].value_counts())
```

## Filtering rows

Imagine you want to know more about just the houses that have a quality score of 10. You can put a condition inside the selection brackets to filter down to just the rows that meet that condition.

```python
ames[ames["Overall Qual"] == 10].describe()
```

## Grouping data

Is overall quality reflective of the price of the house? `groupby()` lets you compress your data based on some criteria. If you pass in a column name, it groups the data by every unique value in that column.

`groupby()` returns a special "groupby object" - you always need to call some aggregation method (like `.mean()`, `.sum()`, `.count()`) on it to actually get a result back. This is often described as **split-apply-combine**: you split your data into groups based on some criteria, apply a function to each group, then combine the results back into a single table.

```python
data_by_qual = ames.groupby("Overall Qual").mean(numeric_only=True)
data_by_qual
```

## Sorting data

Problem: you want to see the stats for the top 10 most expensive houses in the data set. `sort_values()` can sort by one column, or by multiple columns (used as tie-breakers, in order).

```python
ames = ames.sort_values(by=["SalePrice", "Year Built"], ascending=False)
ames.head(10) #passing a number into head will reveal that many rows
```

# More general pandas info

## Series vs. DataFrames

A DataFrame is really just a collection of columns that all share the same index. Each individual column, on its own, is called a **Series**.

```python
single_column = ames["SalePrice"]
type(single_column)
```

```python
#each series has an index and values (like a dictionary)
print(single_column.index)
print()
print(single_column.values)
```

```python
# also noteworthy is that pandas is built on numpy
type(single_column.values)
```

You can also build Series and DataFrames from scratch, out of Python lists and dictionaries. Let's build a small produce inventory as an example.

```python
# from a list
fruit_prices = pd.Series([0.50, 0.25, 1.25], index=["apple", "banana", "mango"])
print(fruit_prices)
print()

# from a dictionary
fruit_stock = {"apple": 120,
               "banana": 75,
               "mango": 40}

fruit_stock = pd.Series(fruit_stock) #keys become index, values are values
print(fruit_stock)
```

```python
# combining two series that share the same index creates a DataFrame
produce = pd.DataFrame({"Price": fruit_prices,
                         "Stock": fruit_stock})
produce
```

```python
# data frames are also built using numpy arrays
produce.values
```

## Accessing rows and columns

There are a few different ways to reach into a DataFrame depending on whether you want a column, a row by its label, or a row by its position.

```python
# easiest way to think about data frames is a dictionary of series objects
produce["Price"] #returns a series object
```

```python
# chain a second selection to grab a specific value
produce["Price"]["apple"]
```

```python
# .loc[] will grab a row by its index value
produce.loc["banana"]
```

```python
# .iloc will grab a row by its position in the dataframe, zero-indexed
produce.iloc[1] 
```

## Filtering with boolean logic

You can query your DataFrame using boolean conditions - this is one of the most common things you'll do with pandas.

```python
since_2000 = ames[ames["Year Built"] >= 2000]
since_2000.shape
```

```python
# you can combine conditions together as well
new_data_frame = ames[
    (ames["Year Built"] > 2000) &
    (ames["SalePrice"] < 200000) &
    (ames["Bldg Type"] == "1Fam")
]
new_data_frame
```

## Creating new columns

You can create new columns based on the data in existing columns - just assign to a column name that doesn't exist yet.

```python
ames["TotalScore"] = ames["Overall Cond"] + ames["Overall Qual"]
ames.head()
```

## Plotting directly from pandas

pandas also interfaces with matplotlib - calling `.plot()` on a Series or DataFrame will build a matplotlib figure for you automatically. The x-axis defaults to the index, and the y-axis defaults to whatever column(s) you specify.

To see whether total score is correlated with sale price, let's first build a DataFrame with total score as the index.

```python
total_score_df = ames.groupby("TotalScore").mean(numeric_only=True)
total_score_df
```

```python
total_score_df["SalePrice"].plot()
#can add to plot just like for matplotlib
plt.ylabel("Sale Price")
```

```python
#personal opinion - it is easier to work with matplotlib alone than
# plotting from data frames directly - matplotlib interfaces well with pandas
fig, ax = plt.subplots()
ax.plot(total_score_df.index, total_score_df["SalePrice"])
ax.set_xlabel("Total Score")
ax.set_ylabel("Sale Price")
ax.set_xticks(np.arange(2, 20, 2))
```
