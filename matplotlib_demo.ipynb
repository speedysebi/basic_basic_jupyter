## Introduction to Axes and Figures

Matplotlib has two main ways to create plots:

    1. a script-like approach
    2. an object-oriented approach 
    
We are going to walk through the object oriented approach as it will help you make more complicated plots if you ever need them. Additionally, there are a lot of syntactical similarities between the two approaches, so if you know the OO approach you can easily use the scripting approach if it is more convenient.

```python
fig, ax = plt.subplots()
```

A couple definitions:

    1. "Figure" is the outer container of the visual you are creating. Individual plots go inside. 
    2. "Axes" are the individual plots (subplots) that go inside of a figure. 
    
In the example above, plt.subplots() creates both a Figure and Axes (that is empty because we have not added anything to it yet. This is equivalent to calling plt.subplots(1,1) (for 1 row of Axes and 1 column of Axes in the Figure).

If we wanted to, we could create a figure with multiple subplots by doing any of the following:

```python
#2 rows, 1 col
fig, ax = plt.subplots(2,1)
```

```python
#1 row, 2 cols
fig, ax = plt.subplots(1,2)
```

```python
#3 rows, 2 cols
fig, ax = plt.subplots(3,2)
```

If you have multiple Axes in your Figure, you can edit any individual Axes object by accessing the specific element in the "ax" variable, which is actually a numpy array when you have multiple Axes.

```python
fig, ax = plt.subplots()
print(type(ax))
fig, ax = plt.subplots(2,2)
print(type(ax))
print(ax.shape)
plt.close("all") #makes it so figures don't render automatically when the cell is run
```

Specific Axes (subplots) are accessed by grabbing the corresponding position in the array. The top row of the Figure corresponds to the first row in the Array, the second to the second, etc. **Remember: array positions are zero indexed.**

Additionally, the way you access elements in a numpy array has more flexibilitiy than in a multidimensional list:

    -List way: list_name[row][col]
    -Numpy way: array_name[row, col] or list way

```python
ax[0][0].set_title("One way to set a title")
ax[1,1].set_title("Another way to set a title")
fig
```

Note that the title for the subplot in the second row is overlapping with the tic marks for the subplot above it. We can easily fix this! 

```python
fig.subplots_adjust(hspace = .35)
fig
```

Note that just about everything about your plots is customizable, and that there are way too many customization features for us to be able to cover everything in this demo. If you are ever wondering "can I change XXX?" the answer is probably yes!

## Let's go back to a simple 1 Axes figure and actually plot some data

```python
x_axis = [1,2,3,4]
y_axis = [4,3,1,5]
fig, ax = plt.subplots()
ax.plot(x_axis,y_axis) #creates a line plot. x-axis data always comes first
```

This plot on its own is not very informative. Let's add some more details to our existing Figure and Axes objects (meaning we will not recall plt.subplots() but rather reuse our existing "fig" and "ax" variables from the last cell).

```python
#add axis labels
ax.set_xlabel("x-axis label")
ax.set_ylabel("y-axis label")

#add title 
ax.set_title("This is the title")

#display the figure
#Note: if showing the figure is the last line in the cell, you don't need to use the "display()" function
display(fig) 
```

Let's make the figure a little more readable by editing the text size and the tick labels, and also make the line a different color

```python
#adjust text size 
ax.set_xlabel("x-axis label", size = 15)
ax.set_ylabel("y-axis label", size = 15)
ax.set_title("This is the title", size = 18)

#adjust tick labels
ax.set_xticks([1,2,3,4,5])
ax.set_yticks([1,2,3,4,5])

#change line color, multple ways to do this, see https://matplotlib.org/stable/gallery/color/named_colors.html
ax.plot(x_axis, y_axis, c = "r")

fig
```

Now let's add a second line and a legend. 

Note, to do this we are going to need to edit the previous line that was drawn. We could do this programmatically, or we could just create a new plot that has everything we need in a single cell. We used the same Figure/Axes across multiple cells for illustration earlier, but in general you will probably want to keep your plots contained in a single cell.

```python
x_axis = [1,2,3,4]
y_axis = [4,3,1,5]
x_axis2 = [1,2,3,4,5]
y_axis2 = [1,2,3,4,5]

fig, ax = plt.subplots()
#draw first line on ax, adding in a label for the legend
ax.plot(x_axis,y_axis, c = "r", label = "Line 1")
#add a second line on ax with label
ax.plot(x_axis2, y_axis2, c = "g", label = "Line 2")

# add the legend, locations https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.legend.html
ax.legend(fontsize = 20, loc = 5)

#adjust text size 
ax.set_xlabel("x-axis label", size = 15)
ax.set_ylabel("y-axis label", size = 15)
ax.set_title("This is the title", size = 18)

#adjust tick labels
ax.set_xticks([1,2,3,4,5])
ax.set_yticks([1,2,3,4,5])
```

What if we wanted to make this a scatter plot?

```python
fig, ax = plt.subplots()
# changed "plot" (that created the line plot) to "scatter"
ax.scatter(x_axis,y_axis, c = "r", label = "Line 1")
ax.scatter(x_axis2, y_axis2, c = "g", label = "Line 2")

# add the legend, locations https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.legend.html
ax.legend(fontsize = 20, loc = 5)

#adjust text size 
ax.set_xlabel("x-axis label", size = 15)
ax.set_ylabel("y-axis label", size = 15)
ax.set_title("This is the title", size = 18)

#adjust tick labels
ax.set_xticks([1,2,3,4,5])
ax.set_yticks([1,2,3,4,5])
```
