# Lab 3: Birthday Probability - Monte Carlo Simulation

Estimate the probability that at least two people in a group share the same birthday

```python
import random
```

```python
def calculate_birthday_prob(group_size, num_sims):
    """
    Calculates the probability that at least two people
    in a group have the same birthday.
    """
    matches = 0

    for i in range(num_sims):
        birthdays = []

        for j in range(group_size):
            birthday = random.randint(1, 365)
            birthdays.append(birthday)

        if len(birthdays) != len(set(birthdays)):
            matches += 1

    return matches / num_sims


def calculate_min_group(threshold, num_sims=10000):
    """
    Finds the smallest group size that reaches
    the given birthday probability.
    """
    group_size = 2

    while calculate_birthday_prob(group_size, num_sims) < threshold:
        group_size += 1

    return group_size
```
